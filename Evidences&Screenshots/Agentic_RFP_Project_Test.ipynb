```python
import sqlite3
from pathlib import Path

db_path = Path("data/rfp_evaluation.db")

connection = sqlite3.connect(db_path)

tables = connection.execute("""
    SELECT name
    FROM sqlite_master
    WHERE type = 'table'
    ORDER BY name
""").fetchall()

print("Database tables:")

for table in tables:
    print("-", table[0])

connection.close()
```

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("data/rfp_evaluation.db")

criteria = pd.read_sql_query("""
    SELECT
        criterion_id,
        name,
        description,
        weight,
        max_score,
        is_active
    FROM evaluation_criteria
    ORDER BY criterion_id
""", connection)

connection.close()

display(criteria)

print("Active weight total:", criteria.loc[criteria["is_active"] == 1, "weight"].sum())
```

```python
import json

with open("exports/sample_completed_run.json", "r", encoding="utf-8") as file:
    successful_run = json.load(file)

print("Run ID:", successful_run["rfp_run_id"])
print("Status:", successful_run["status"])
print("Number of suppliers:", len(successful_run["suppliers"]))

for supplier in successful_run["suppliers"]:
    print(
        supplier["final_rank"],
        supplier["supplier_name"],
        "Absolute Score:",
        supplier["absolute_score"],
        "PPI:",
        supplier["ppi"]
    )
```

```python
for supplier in successful_run["suppliers"]:
    print("\nSupplier:", supplier["supplier_name"])

    for row in supplier["scorecard"]:
        print(
            row["criterion_name"],
            "| Score:", row["score"],
            "| Benchmark:", row["benchmark"],
            "| Gap:", row["gap"],
            "| Relative %:", round(row["relative_percent"], 2),
            "| Weight:", row["weight"]
        )
```

```python
with open("exports/validation_error_case.json", "r", encoding="utf-8") as file:
    validation_case = json.load(file)

print("Original input:")
print(json.dumps(validation_case["input"], indent=2))

print("\nWarnings:")
for warning in validation_case["warnings"]:
    print("-", warning)

print("\nNormalized output:")
print(json.dumps(validation_case["normalized_output"], indent=2))
```
